# Building AI Apps with Lakebase-Powered Memory
#### Duration: 5 minutes

## Overview

In this hands-on lab, you will build a complete stateful AI assistant for **Bakehouse**, a global bakery franchise chain with 48 locations across 9 countries. Starting from live sales data, you will give the assistant tools to answer business questions, add both short-term and long-term memory so it can hold a conversation and remember each manager across sessions, and finally deploy it as a production Databricks App with a chat interface.

The memory that makes this possible is powered by **Lakebase**, Databricks' PostgreSQL-compatible operational database. You will use LangGraph with the `databricks-langchain` CheckpointSaver and DatabricksStore to persist conversation state and curated facts, then wire the same memory systems into an async FastAPI app for production use.

## Terminal Objectives

By the end of this lab, you will be able to:

- **Create Unity Catalog Function tools** that query live `samples.bakehouse` sales data and register them so a LangGraph agent can call them.
- **Provision a Lakebase instance** and use it as the operational backend for agent memory.
- **Implement short-term memory** with a LangGraph CheckpointSaver so the agent maintains context across a multi-turn conversation.
- **Implement long-term memory** with a DatabricksStore so the agent recalls user-specific facts across separate sessions.
- **Assemble and test a stateful LangGraph agent** that combines UC Function tools with both memory systems.
- **Deploy the agent as a Databricks App** using FastAPI, async memory, and a branded chat interface.
- **Inspect both memory systems in Lakebase** and compare the self-managed pattern with Managed Agent Memory (Beta).

## A. Prerequisites

Before starting this course, learners should be comfortable with the following:

- **General familiarity with the Databricks workspace** - You can navigate the workspace UI, select compute, and run notebooks.

- **Familiarity with Unity Catalog concepts** - You understand catalogs, schemas, tables, and functions in Unity Catalog. Prior experience registering UC Functions is helpful but not required.

- **Intermediate Python proficiency** - You can read and write Python, work with functions and dictionaries, and follow object-oriented code. Prior exposure to LangChain or LangGraph is helpful but not required.

- **Intermediate SQL proficiency** - You can read and write SQL queries including SELECT, JOIN, GROUP BY, and aggregate functions.

- **Basic familiarity with generative AI agents** - You understand what an LLM agent is and the idea of giving an agent tools. Prior experience building agents is not required.

- **Access to a Databricks workspace** with permission to select Serverless Compute, create a Lakebase instance, and deploy a Databricks App.
